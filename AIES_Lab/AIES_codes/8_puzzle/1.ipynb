```python
from collections import deque

# Function to print puzzle
def print_state(state):
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()
```

```python
# Get index of blank (0)
def get_blank_pos(state):
    return state.index(0)

# Generate all possible moves
def get_neighbors(state):
    neighbors = []
    blank = get_blank_pos(state)

    moves = {
        "UP": -3,
        "DOWN": 3,
        "LEFT": -1,
        "RIGHT": 1
    }

    for move in moves:
        new_state = state.copy()
        new_blank = blank + moves[move]

        # Conditions to avoid invalid moves
        if move == "LEFT" and blank % 3 == 0:
            continue
        if move == "RIGHT" and blank % 3 == 2:
            continue
        if new_blank < 0 or new_blank >= 9:
            continue

        # Swap
        new_state[blank], new_state[new_blank] = new_state[new_blank], new_state[blank]
        neighbors.append(new_state)

    return neighbors

```

```python
# BFS to generate state space
def generate_state_space(initial, goal):
    visited = set()
    queue = deque([initial])

    while queue:
        current = queue.popleft()

        # Convert list to tuple for set
        state_tuple = tuple(current)

        if state_tuple in visited:
            continue

        visited.add(state_tuple)

        print("Current State:")
        print_state(current)

        # Stop if goal reached
        if current == goal:
            print("Goal Reached!")
            return

        # Add neighbors
        for neighbor in get_neighbors(current):
            queue.append(neighbor)

# ---- INPUT ----
initial = list(map(int, input("Enter initial state (9 numbers with space): ").split()))
goal = list(map(int, input("Enter goal state (9 numbers with space): ").split()))

generate_state_space(initial, goal)
```
